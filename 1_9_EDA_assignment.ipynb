```python
# Task 1: Time Series Setup
import pandas as pd
import numpy as np

# ── 1. Create DataFrame with daily dates for 2023 ──────────────────────────
date_index = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')

# ── 2. Generate random Sales data ──────────────────────────────────────────
np.random.seed(42)
sales = np.random.randint(100, 1000, size=len(date_index))

df = pd.DataFrame({'Sales': sales}, index=date_index)
print("Daily Sales (first 7 days):")
print(df.head(7))

# ── 3. Resample to Monthly Total Sales ─────────────────────────────────────
monthly_sales = df.resample('ME').sum()
print("\nMonthly Total Sales:")
print(monthly_sales)
```

```python
import pandas as pd
import numpy as np

# Task 2: The "Wide" Report
sales = np.random.randint(100, 1000, size=15)
df = pd.DataFrame({
    'Date': ['2023-01', '2023-02', '2023-03', '2023-04', '2023-05']*3,
    'Region': ['North', 'South', 'East']*5,
    'Sales': sales
})

print("Original DataFrame:\n", df)

df_pivot = df.pivot(index='Date', columns='Region', values='Sales')
print("\nPivoted DataFrame:\n", df_pivot)

df_pivot = df_pivot[['North', 'South', 'East']]
print("\n", df_pivot)
```

```python
# Task 3: The Executive Summary
tip = pd.read_csv('tips.csv')

print("Original table:\n", tip.head(5))

day_order = ['Thur', 'Fri', 'Sat', 'Sun']
tip_pivot = tip.pivot_table(index='day', columns='time', values='total_bill', aggfunc='mean', margins=True, margins_name='All').loc[day_order + ['All']]
print("Pivoted table: \n", tip_pivot.round(2))
```
